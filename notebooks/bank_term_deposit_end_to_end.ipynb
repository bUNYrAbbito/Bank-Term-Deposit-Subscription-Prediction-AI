# Bank-Term Deposit Subscription Prediction
### End-to-End Machine Learning, Advanced Ensembling, Explainable AI (SHAP), and Deployment

**Domain**: Financial Services, Banking, Predictive Analytics  
**Objective**: Build a binary classification model that accurately predicts whether a client will subscribe to a term deposit offer based on demographic, financial, and campaign outreach features.

---
### Key Skills & Capabilities Covered:
1. **Tabular Data Preprocessing & Feature Engineering**: Ordinal encoding of ordered categories, one-hot encoding of nominal variables, robust scaling of skewed distributions, interaction features (duration/campaign, balance/age, debt indices, cyclical date encodings).
2. **Handling Imbalanced Classes**: Class-weight balancing, stratified sampling, and decision threshold calibration.
3. **Multi-Architecture Modeling**:
   - Baselines: Logistic Regression, Gaussian Naive Bayes, Support Vector Machines (SGD with log-loss)
   - Tree & Ensemble: Random Forest
   - Gradient Boosting: XGBoost, CatBoost, Histogram Gradient Boosting
   - Deep Learning: Multi-Layer Perceptron (MLP with ReLU and Adam)
   - Meta-Ensembles: Stacking Classifier and Soft Voting Ensemble
4. **Validation & Zero Data Leakage**: Stratified 5-Fold Cross-Validation, transformers fitted solely on training folds.
5. **Model Evaluation**: Accuracy, Precision, Recall, F1-Score, ROC-AUC, PR-AUC, Confusion Matrices, and Cost-Benefit Threshold Tuning.
6. **Model Explainability**: Gini/Gain Feature Importances, SHAP Beeswarm summary, SHAP global importance, and SHAP local waterfall explanations.
7. **Production Deployment**: Interactive Streamlit Dashboard (`app.py`), Docker containerization (`Dockerfile`, `docker-compose.yml`), and AWS deployment guide.
